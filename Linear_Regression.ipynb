# Linear Regression Basics:

It is like a straight line relationship between Inputs and Outputs.

Line follows a formula : y = mx + c , where
      
      y = Predicted feature

      x = Input features

      m(alpha) = slope of the relationship

      c(beta) = intercept means where the line starts.

To make the best prediction we need to find the value of alpha and beta so that the line fits the data as closely as possible.

# Linear Regression Mathematically:

Take the mean of x and y axis data called as x and y bar. and plot the intersection.

Then find the x - xbar and y - ybar and square of them to ignore the nullification. Then multiply x - xbar and y - y bar.

m = sum (x - xbar into y - ybara) / sum sq.(x - xbar)

For value of c we have to use the formula ybar = m xbar + c

```python
import numpy as np
```

```python
np.random.seed(50)
```

```python
X = np.random.rand(50,1)*100
```

```python
Y = 3.5 * X + np.random.randn(50, 1) * 20
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model.fit(X, Y)
```

```python
Y_pred = model.predict(X)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8,6))
plt.scatter(X, Y, color='blue', label='Data Points')
plt.plot(X, Y_pred, color='red', linewidth=2, label='Regression Line')
plt.title('Linear Regression on Random Dataset')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)
plt.show()
```

```python
print("Slope (Coefficient):", model.coef_[0][0])
print("Intercept:", model.intercept_[0])
```
