```As we can see that there are multiple strings and our computer is not able to underestand them so we need to convert them into the integer.```

```There are two types of encoding strategy : for this conversion.```

```1. Label Encoding: if strings are in order or ranking they are assigned with sequential numbers from 0 to n-1.```

```Ex: 0 = Iris-setosa, 1 = Iris-versicolor, 2 = Iris-virginica.```

```Best use for tree-based machine learning algorithms like Random Forest, Decision Trees, or XGBoost which handel sequential number well.```

```2. One Hot Encoding: if strings have no logical order, then assigning sequential integers can mislead the model. Instead, the column is split into multiple separate columns and only assign 0 and 1.```

```Ex: This converts the single column into three separate columns. Each column only contains a 0 (absent) or 1(present).```

```like: [1,0,0] = Iris-setosa, [0,1,0] = Iris versicolor, [0,0,1] = Iris-virginica.```

```Best used for Distance based algorithms like Logistic Regression, SVM, Neural Networks.```

**```We can perform the similar task using mapping.```**

````python
mapping = {'Iris-setosa' : 0, 'Iris-versicolor' : 1, 'Iris-virginica' : 2}
````

````python
df['label'] = df['label'].map(mapping)
````
